```python
import json
from pathlib import Path
import pandas as pd
```

```python
GOLDEN_DIR = Path(r"C:\Users\Satyajit_Khot\Downloads\techfiling-copilot\goldens")
def load_dataset(name):
    return [json.loads(l) for l in (GOLDEN_DIR / f"{name}").open(encoding="utf-8")]
```
